## **Testing for arguments of RQ3 PART 1**

```python
from src.data_loader import (
    load_transactions_combined_simple,
    load_transactions_graph_gcn,
)

# ---------------------------------------------------
# 1. Inspect the FEATURE-BASED dataset (RQ1 setup)
# ---------------------------------------------------
X_train_feat, X_test_feat, y_train_feat, y_test_feat = load_transactions_combined_simple()

print("=" * 60)
print("FEATURE-BASED DATASET (elliptic-transactions-combined)")
print("=" * 60)
print(f"Total features: {X_train_feat.shape[1]}")
print(f"Train rows: {X_train_feat.shape[0]}, Test rows: {X_test_feat.shape[0]}")
print(f"\nFirst 20 column names:")
for col in list(X_train_feat.columns)[:20]:
    print(f"  - {col}")
print(f"\nLast 20 column names:")
for col in list(X_train_feat.columns)[-20:]:
    print(f"  - {col}")
```

```python
# ---------------------------------------------------
# 2. Inspect the GRAPH dataset (RQ2 setup)
# ---------------------------------------------------
graph_data = load_transactions_graph_gcn()

print("=" * 60)
print("GRAPH DATASET (elliptic-transactions-graph)")
print("=" * 60)
print(f"Total node features: {graph_data.x.shape[1]}")
print(f"Total nodes: {graph_data.x.shape[0]}")
print(f"Total edges: {graph_data.edge_index.shape[1]}")
```

```python
# ---------------------------------------------------
# 3. Verify both representations use the SAME features
# ---------------------------------------------------
print("=" * 60)
print("FAIR COMPARISON CHECK")
print("=" * 60)
print(f"Feature-based feature count: {X_train_feat.shape[1]}")
print(f"Graph node feature count:    {graph_data.x.shape[1]}")
print(f"Match: {X_train_feat.shape[1] == graph_data.x.shape[1]}")
```

```python
# ---------------------------------------------------
# 4. Check feature composition - count aggregated vs local features
# ---------------------------------------------------
all_cols = list(X_train_feat.columns)

# Elliptic++ aggregated feature naming patterns
aggregated_keywords = [
    "in_BTC", "out_BTC",        # neighbour Bitcoin aggregates
    "in_txs", "out_txs",        # neighbour transaction counts
]

aggregated_cols = [
    col for col in all_cols
    if any(kw in col for kw in aggregated_keywords)
]
local_cols = [col for col in all_cols if col not in aggregated_cols]

print("=" * 60)
print("FEATURE COMPOSITION")
print("=" * 60)
print(f"Total features:      {len(all_cols)}")
print(f"Local features:      {len(local_cols)}")
print(f"Aggregated features: {len(aggregated_cols)}  ({len(aggregated_cols)/len(all_cols)*100:.1f}%)")

print(f"\nAggregated columns found:")
for col in aggregated_cols:
    print(f"  - {col}")
```

```python
# Compute average node degree
n_nodes = graph_data.x.shape[0]
n_edges = graph_data.edge_index.shape[1]
avg_degree = n_edges / n_nodes

print(f"Total nodes: {n_nodes:,}")
print(f"Total edges: {n_edges:,}")
print(f"Average degree: {avg_degree:.2f}")
```
