```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent  # because notebook is in /notebooks
DATA = ROOT / "data"

def check_csv(path: Path, n=3):
    print("\n" + "="*80)
    print("FILE:", path)
    print("EXISTS:", path.exists())
    if not path.exists():
        return
    df = pd.read_csv(path)
    print("SHAPE:", df.shape)
    print("COLUMNS (first 10):", list(df.columns[:10]))
    print("HEAD:")
    display(df.head(n))
```

```python
check_csv(DATA / "elliptic_plus_plus" / "txs_features.csv")
check_csv(DATA / "elliptic_plus_plus" / "txs_classes.csv")
check_csv(DATA / "elliptic_plus_plus" / "txs_edgelist.csv")
check_csv(DATA / "elliptic_plus_plus" / "AddrTx_edgelist.csv")
check_csv(DATA / "elliptic_plus_plus" / "TxAddr_edgelist.csv")
check_csv(DATA / "elliptic_plus_plus" / "AddrAddr_edgelist.csv")
check_csv(DATA / "elliptic_plus_plus" / "wallets_features.csv")
check_csv(DATA / "elliptic_plus_plus" / "wallets_classes.csv")
```
