```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent
DATA = ROOT / "data" / "elliptic_plus_plus"
```

```python
features = pd.read_csv(DATA / "wallets_features.csv")
classes = pd.read_csv(DATA / "wallets_classes.csv")
combined = pd.read_csv(DATA / "wallets_features_classes_combined.csv")
```

```python
print("Features rows:", len(features))
print("Classes rows:", len(classes))
print("Combined rows:", len(combined))
```

```python
key = "address"
```

```python
missing_from_combined = set(features[key]) - set(combined[key])
print("Missing from combined (features):", len(missing_from_combined))
```

```python
missing_from_combined_classes = set(classes[key]) - set(combined[key])
print("Missing from combined (classes):", len(missing_from_combined_classes))
```

```python
expected_columns = set(features.columns).union(set(classes.columns))
combined_columns = set(combined.columns)

missing_columns = expected_columns - combined_columns
print("Missing columns in combined:", missing_columns)
```

```python
merged_check = pd.merge(features, classes, on=key, how="inner")
```

```python
print("Your merge shape:", merged_check.shape)
print("Given combined shape:", combined.shape)
```

```python
# Sort for fair comparison
merged_check_sorted = merged_check.sort_values(by=key).reset_index(drop=True)
combined_sorted = combined.sort_values(by=key).reset_index(drop=True)

# Check equality
print(merged_check_sorted.equals(combined_sorted))
```
