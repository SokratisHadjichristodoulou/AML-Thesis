```python
from src.data_loader import load_transactions_graph_baseline_simple

X_train, X_test, y_train, y_test, train_time, test_time = load_transactions_graph_baseline_simple()

assert "Time step" not in X_train.columns
assert "Time step" not in X_test.columns
assert len(test_time) == len(y_test) == len(X_test)
assert len(train_time) == len(y_train) == len(X_train)
assert max(train_time) < min(test_time), "Time boundary not clean!"

print(f"Train steps: {sorted(train_time.unique())[0]}–{sorted(train_time.unique())[-1]} ({len(train_time)} rows)")
print(f"Test steps:  {sorted(test_time.unique())[0]}–{sorted(test_time.unique())[-1]} ({len(test_time)} rows)")
print(f"Illicit rate — train: {y_train.mean():.4f}, test: {y_test.mean():.4f}")
```
